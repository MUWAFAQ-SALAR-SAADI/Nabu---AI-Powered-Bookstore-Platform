```python
import numpy as np

emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]
isbn = []
emotion_scores = {label: [] for label in emotion_labels}

def calculate_max_emotion_scores(predictions):
    per_emotion_scores = {label: [] for label in emotion_labels}
    for prediction in predictions:
        sorted_predictions = sorted(prediction, key=lambda x: x["label"])
        for index, label in enumerate(emotion_labels):
            per_emotion_scores[label].append(sorted_predictions[index]["score"])
    return {label: np.max(scores) for label, scores in per_emotion_scores.items()}
```

```python
from transformers import pipeline
from collections import defaultdict, Counter

# إنشاء المصنّف
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=-1  # CPU، إذا عندك GPU استخدم device=0
)

# مثال على الفئات / المشاعر التي تريد التنبؤ بها
emotion_labels = ["joy", "sadness", "anger", "fear", "surprise", "disgust"]

# إعداد القوائم لتخزين النتائج
isbn = []
emotion_scores = defaultdict(list)

# دالة لحساب أعلى الاحتمالات لكل فئة
def calculate_max_emotion_scores(predictions):
    max_scores = {label: 0.0 for label in emotion_labels}
    for pred in predictions:
        for label, score in zip(pred['labels'], pred['scores']):
            if score > max_scores[label]:
                max_scores[label] = score
    return max_scores

# تصنيف أول 10 كتب كمثال
for i in range(10):
    isbn.append(books["isbn13"][i])
    sentences = [s.strip() for s in books["description"][i].split(".") if s.strip()]

    # تصنيف كل جملة مع تحديد الفئات
    predictions = [
        classifier(sentence, candidate_labels=emotion_labels)
        for sentence in sentences
    ]

    # حساب أعلى الاحتمالات لكل فئة
    max_scores = calculate_max_emotion_scores(predictions)

    # تخزين النتائج
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])

# عرض النتائج
import pandas as pd
df_emotions = pd.DataFrame(emotion_scores)
df_emotions["isbn13"] = isbn
print(df_emotions)
```

```python
from tqdm import tqdm
from transformers import pipeline
import pandas as pd
import os

# إعداد المصنّف
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=-1  # CPU
)

# فئات المشاعر
emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]

# دالة لحساب أعلى الاحتمالات لكل فئة
def calculate_max_emotion_scores(predictions):
    max_scores = {label: 0.0 for label in emotion_labels}
    for pred in predictions:
        for label, score in zip(pred['labels'], pred['scores']):
            if score > max_scores[label]:
                max_scores[label] = score
    return max_scores

# لو فيه ملف نتائج سابق نكمل منه
save_file = "emotion_scores_partial.csv"
if os.path.exists(save_file):
    df_existing = pd.read_csv(save_file)
    processed_isbns = set(df_existing["isbn13"])
    emotion_scores = df_existing.to_dict(orient="list")
    isbn = emotion_scores.pop("isbn13")
else:
    df_existing = None
    processed_isbns = set()
    isbn = []
    emotion_scores = {label: [] for label in emotion_labels}

# معالجة الكتب
for i in tqdm(range(len(books))):
    book_id = books["isbn13"][i]
    if book_id in processed_isbns:
        continue  # هذا الكتاب معالج مسبقاً

    sentences = [s.strip() for s in books["description"][i].split(".") if s.strip()]

    # تصنيف كل الجمل دفعة واحدة
    predictions = classifier(sentences, candidate_labels=emotion_labels, multi_label=True)

    # حساب أعلى الاحتمالات لكل فئة
    max_scores = calculate_max_emotion_scores(predictions)

    isbn.append(book_id)
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])

    # حفظ كل 100 كتاب
    if len(isbn) % 100 == 0:
        df_emotions = pd.DataFrame(emotion_scores)
        df_emotions["isbn13"] = isbn
        df_emotions.to_csv(save_file, index=False)
        print(f"✅ Progress saved at {len(isbn)} books")

# الحفظ النهائي
df_emotions = pd.DataFrame(emotion_scores)
df_emotions["isbn13"] = isbn
df_emotions.to_csv(save_file, index=False)
print("🎉 Finished and saved all results")
```
